### The goal of this turorial is to implement Logistic Regression from scratch

> Main Things one needs to know while implementing an ML algorithm

* Model y = f(X)
  * For Logistic Regression - Model = Sigmoid(f(x)) = Sigmoid(X*W + bias)
  * Sigmoid(z) = S(z) = exp(z)/1+exp(z) = 1/1+exp(-z) where z = X*W+bias
---  
* Derivative of a Sigmoid is S'(z) = S(z)(1-S(z))
---
* Cost or Loss function for the model.
  * (-1/num_samples)[(y*log(prediction) + (1-y)*log(1-prediction)] where prediction = S(z)  
---      
* What variables in f(X) is our algorithm learning?
  * For logrithmic regression the algorithm is learning slope (W; weights) and intercept (b; bias)
---      
* What learning algorithm is being used ?
  * Gradient Descent.
---      
* Since we are using Gradient Descent - it becomes very important to understand the derivative of the loss function with respect to the variables one is trying to learn.
  
    * derivative of loss function wrt weights =  \
         (-1/num_samples)*(A-Y)X where A = prediction = S(z)
      
    * derivative of loss function wrt bias =    (-1/num_samples)*(A-Y) where A = S(z)
  
---  
* <b> When you program  the algorithm from scratch pay close attention to dimensionality at every step.</b>


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    """ Function returns sigmoid of z"""
    return np.exp(z)/(1+np.exp(z))
```

> Illustration of a Sigmoid function

```python
x =  np.arange(-5, 5, 0.1)
y = sigmoid(x)
x1 = np.arange(-5,5,0.05)
z = 3.5*x1+0.86
y1 = sigmoid(z)


plt.plot(x,y,'r--.')
plt.plot(x1,y1,'g--.')
plt.title('Illustration of a sigmoid function')
plt.xlabel('X')
plt.ylabel('Sigmoid')
plt.grid()
```
